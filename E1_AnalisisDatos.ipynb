## 4. Análisis de los Descriptores Numéricos

### 4.1 Rango de valores de cada descriptor

Vamos a ver el rango (máximo - mínimo) de cada variable para entender en qué escala
se mueve cada una.

```python
descriptores = ['SL', 'EEG', 'BP', 'HR', 'CIRCULATION']

print("Rango de cada descriptor:")
print(traindf[descriptores].max() - traindf[descriptores].min())
```

Las diferencias de escalas entre variables son muy grandes.
* `SL` y `EEG` se mueven en rangos de millones
* `CIRCULATION` en decenas de miles
* `BP` y `HR` tienen rangos de apenas cientos.

Esto nos indica que antes de usar modelos
sensibles a la escala como SVM, KNN o redes neuronales vamos a tener que escalar los datos

### 4.2 Valores extremos y anomalías

Podemos ordenar el dataset por el valor de un descriptor. Esto resulta util para buscar los extremos.

Se puede hacer con `sort_values` o con `apply`




```python
traindf.sort_values(by='EEG', ascending=False).head()
```

```python
traindf.sort_values(by='HR', ascending=True).head()
```

```python
traindf[descriptores].apply([np.min, np.max])
```

Viendo los extremos de todos los descriptores podemos detectar algo:
- `EEG` tiene valores negativos
- `BP` tiene un mínimo de 0 que es fisiológicamente imposible, por lo que vamos a verlo un poco mas a detalle


```python
print(f"Registros con BP = 0: {(traindf['BP'] == 0).sum()}")
traindf[traindf['BP'] == 0]['ACTIVITY'].value_counts()
```

Hay 17 registros con BP = 0, lo que representa apenas el 0.14% del total. Al ver su distribución por clase, se reparten entre varias actividades
distintas, lo que descarta que sea una característica propia de alguna actividad concreta. Al ser tan pocos lo más probable es que
sean errores de medición.

### 4.3 Estadísticas por clase de actividad
Lo más interesante es ver cómo varian los descriptores según la actividad que se está realizando. Para eso, se calculan la media `mean()` y la desviación típica `std()` de cada descriptor en los datasets. Si esos valores varían bastante entre clases, la señal tiene potencial para ayudar al clasificador a distinguir entre ellas; si son prácticamente iguales para todas las actividades, probablemente no aporte mucho.

```python
traindf.groupby('ACTIVITY')[descriptores].mean()
```

```python
traindf.groupby('ACTIVITY')[descriptores].std()
```

* **Clase 0** tiene los valores más altos en `SL` (115 259.18), `HR` (259.14), `BP` (72.43)y `CIRCULATION` (4 035.96). En `EEG` (-10 439.16) tiene valores parecidos al resto. Es la clase con los valores de la mayoría de los descriptores más altos con respecto a las otras clases, por lo que será probablemente una clase fácil de identificar.

* **Clase 1** tiene los valores más bajos en todos los descriptores con `SL` (18 772.61),`HR` (117.02), `BP` (39.54) y `CIRCULATION` (927.05) y con el `EEG` menos negativo de todos (-4 347.17). Esa combinación la hace bastante identificable del resto.

* **Clase 2** se diferencia casi exclusivamente por su `EEG` extremadamente negativo (-60 734.70), muy por debajo del resto de clases. Sin embargo, su desviación típica en `EEG` es la mayor de toda la tabla (218 980.92), lo que nos dice que hay gran dispersión interna que puede complicar su identificación y clasificación.

* **Clases 3 y 4** son muy parecidas entre sí: `EEG` prácticamente idéntico
(-14 340.71 vs -14 377.42). `BP` (51.20 vs 57.16), `HR` (217.31 vs 199.95) y
`CIRCULATION` (2 946.51 vs 2 613.27) son descriptores que, considerando la
desviación, solapan mucho sus valores. El `SL` sí cambia algo (74 155.77 vs
63 937.62), pero sus desviaciones (96 383.35 y 105 893.79) son tan grandes que
esa diferencia es difícil de detectar.

* **Clase 5** tiene un `EEG` algo más negativo (-35 512.77) que las clases 3 y 4, lo que la aleja ligeramente, pero su desviación en `EEG` (148 018.76) es muy alta y no lo convierte en un separador fiable. En conjunto, las clases 3, 4 y 5 serán la fuente principal de errores del clasificador.

## 5. Preguntas sobre el Dataset


### Pregunta 1: ¿Es verdad que la media de HR de la clase 0 es superior a la media de HR del resto de clases combinadas?

```python
traindf[traindf['ACTIVITY'] == 0]['HR'].mean()
```

```python
traindf[traindf['ACTIVITY'] != 0]['HR'].mean()
```

Si, es verdad. La media de HR de la clase 0 es bastante superior al la del resto de clases combinadas.( 259.15 vs 194.37 )

### Pregunta 2: ¿Cuántos registros de la clase 2 tienen EEG por debajo de -100.000?

En el análisis anterior vimos que la clase 2 tiene el EEG más negativo y la mayor desviación típica. Queremos saber cuántos registros son realmente extremos.

```python
clase2 = traindf[traindf['ACTIVITY'] == 2]
extremos = clase2[clase2['EEG'] < -100000]
```

```python
print(len(extremos))
print(f'{len(extremos) / len(clase2) * 100:.2f}%')
```

Solo 122 registros de la clase 2 tienen EEG por debajo de -100.000, que es solo el 6.60% de esa clase. Esto significa que aunque la media de EEG de la clase 2 es la mas negativa, la  mayoria de sus registros no son tan extremos. La desviación típica que observamos en el apartado 4.3 se debe a unos pocos valores muy extremos, lo que cambia nuestra anterior suposicion que identificar la clase 2 seria algo mas sencillo.

### Pregunta 3: ¿Cuántos registros de la clase 4 tienen valores de SL similares a de los de la clase 3?
Comprobamos si los registros de clase 3 y 4 se solapan para el descriptor SL.

```python
q25 = traindf[traindf['ACTIVITY'] == 3]['SL'].quantile(0.25)
q75 = traindf[traindf['ACTIVITY'] == 3]['SL'].quantile(0.75)
c4 = traindf[
    (traindf['ACTIVITY'] == 4) &
    (traindf['SL'] >= q25) &
    (traindf['SL'] <= q75)]
```

```python
print(f'Rango : [{q25:.2f}, {q75:.2f}]')
print(len(c4))
print(f'Porcentaje:{len(c4) / len(traindf[traindf["ACTIVITY"]==4]) * 100:.1f}%')
```

El 53.6% de los registros de la clase 4 caen dentro del rango de la clase 3 [12.048, 90.469] del descriptor SL

###Pregunta 4: ¿Los registros con CIRCULATION muy alta (> percentil 90) de que clase son mayoritariamente?

```python
q90 = traindf['CIRCULATION'].quantile(0.90)
preg4 = traindf[traindf['CIRCULATION'] > q90]
```

```python
print(f'Total: {len(preg4)}')
```

```python
print(preg4['ACTIVITY'].value_counts())
```

De los 1170 registros con CIRCULATION por encima del percentil 90, 598 pertenecen a la clase 0, lo que representa el 51.1% del total. Esto confirma lo que ya veiamos con las medias, la clase 0 tiene la mayoría de los valores extremos de CIRCULATION

### Pregunta 5: ¿Cuántos registros tienen HR superior a 300? ¿A qué clase pertenecen mayoritariamente?

```python
preg5 = traindf[traindf['HR'] > 300]
```

```python
print(len(preg5))
print(preg5['ACTIVITY'].value_counts())
```

Hay 2364 registros con HR superior a 300, de ellos 1075 pertenecen a la clase 0, siendo con diferencia la clase mayoritaria en este grupo.
